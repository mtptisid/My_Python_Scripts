```python
import pypdf 

with open("myfile.pdf", "rb") as f:
    reader = pypdf.PdfReader(f)
    pages = reader.pages

    print(f"Total pages: {len(pages)}")

    for i, page in enumerate(pages):
        print(f"Page {i + 1}:")
        print(page.extract_text())
```

```python
import pdfplumber

with pdfplumber.open("myfile.pdf") as pdf:
    pages = pdf.pages

    print(f"Total Pages: {len(pages)}")

    for i, page in enumerate(pages):
        print(f"\n--- Page {i + 1} ---")
        text = page.extract_text()
        print(text)
```

```python
table = page.find_table()

if table:
     data = table.extract()
     for row in data:
         print(row)
else:
    print("No table found on this page.")
```

```python
for rows in table.rows:
    print(rows.cells)
```
